```python
# 环境准备: 让 notebook 能 import 到项目根的 src 包
import sys, pathlib
ROOT = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == 'notebooks' else pathlib.Path.cwd()
sys.path.insert(0, str(ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_data
from src.preprocess import describe_data

pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
print('项目根目录:', ROOT)
```

```python
# 1. 读取数据
# 有真实数据时: df, source = load_data(ROOT / 'data/raw/你的数据.csv')
df, source = load_data(demo=True)   # 先用演示数据跑通, 再换成真实数据
print('数据来源:', source)
print('数据规模:', df.shape)
df.head()
```

```python
# 2. 数据概览: 类型 / 缺失 / 唯一值
describe_data(df)
```

```python
# 3. 缺失值可视化
missing = df.isna().mean().sort_values(ascending=False) * 100
print('缺失率 Top10 (%):')
print(missing.head(10))

if missing.max() > 0:
    plt.figure(figsize=(8, 4))
    missing[missing > 0].plot(kind='bar')
    plt.title('各列缺失率 (%)')
    plt.ylabel('%')
    plt.show()
else:
    print('该数据集没有缺失值。')
```

```python
# 4. 数值特征描述性统计与相关性
num = df.select_dtypes(include=[np.number])
display(num.describe().T)

if num.shape[1] >= 2:
    plt.figure(figsize=(max(6, num.shape[1]), max(5, num.shape[1] * 0.8)))
    sns.heatmap(num.corr(), annot=num.shape[1] <= 12, fmt='.2f', cmap='coolwarm', center=0)
    plt.title('相关性热力图')
    plt.show()
```

```python
# 5. 目标变量分布（按你的数据集修改 target 列名）
target = 'target'
if target in df.columns:
    plt.figure(figsize=(6, 4))
    if df[target].nunique() <= 20:
        sns.countplot(x=df[target])
    else:
        sns.histplot(df[target], kde=True)
    plt.title(f'目标变量分布: {target}')
    plt.show()
else:
    print('请把 target 改成你数据集的目标列名, 现有列:', list(df.columns))
```
